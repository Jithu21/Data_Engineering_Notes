```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

spark = SparkSession.builder.appName("pyspark_assignment").master("local[*]").getOrCreate()
print(spark)

sc = spark.sparkContext
```

Question 1: RDD – Filter and Transform Numbers

```python
numbers = [5, 12, 7, 20, 3, 16, 9, 24, 10, 1]

rdd = spark.sparkContext.parallelize(numbers)
```

```python
rdd_1 = rdd.filter(lambda x: x > 10)
rdd_1.collect()
```

```python
rdd_2 = rdd_1.map(lambda x: x * 2)
rdd_2.collect()
```

```python
final_rdd = rdd_2
final_rdd.collect()
```

Question 2: RDD – Student Marks



```python
student_marks = [
    ("Arun", 78),
    ("Divya", 92),
    ("Kavin", 45),
    ("Meena", 66),
    ("Ravi", 35),
    ("Priya", 88)
]

rdd_student = spark.sparkContext.parallelize(student_marks)
rdd_student.collect()
```

Question 3: DataFrame – Select and Filter Employees

```python
employee_data = [
    (1, "Arun", "Male", 24, "Chennai", 28000),
    (2, "Divya", "Female", 27, "Bengaluru", 42000),
    (3, "Kavin", "Male", 22, "Chennai", 25000),
    (4, "Meena", "Female", 30, "Hyderabad", 52000),
    (5, "Ravi", "Male", 28, "Bengaluru", 38000),
    (6, "Priya", "Female", 25, "Chennai", 46000)
]

columns = ["id", "name", "gender", "age", "city", "salary"]

emp_df = spark.createDataFrame(employee_data, columns)
emp_df.show()
```

```python
new_emp_df = emp_df.select("name", "city", "salary")
new_emp_df.show()
```

```python
final_emp_df = new_emp_df.where("salary > 40000")
final_emp_df.show()
```

Question 4: DataFrame – Add a Column Using "withColumn()"

```python
product_data = [
    (101, "Notebook", 40, 5),
    (102, "Pen", 10, 20),
    (103, "Pencil", 5, 30),
    (104, "School Bag", 650, 3),
    (105, "Eraser", 8, 15)
]

columns = ["product_id", "product_name", "price", "quantity"]

product_df = spark.createDataFrame(product_data, columns)
product_df.show()
```

```python
product_df_1 = product_df.withColumn("total_value", expr("price *quantity"))

product_df_1.show()
```

```python
final_product_df = product_df_1
final_product_df.show()
```

Question 5: DataFrame – Filter, Select, and Add a Column

```python
movie_data = [
    ("Leo", "Action", 250, 7.8),
    ("Hi Nanna", "Drama", 75, 8.2),
    ("Jailer", "Action", 200, 7.1),
    ("Good Night", "Comedy", 30, 8.0),
    ("Dragon", "Comedy", 100, 7.5),
    ("Amaran", "Drama", 150, 8.3)
]

columns = ["movie_name", "category", "collection_crore", "rating"]
```

```python
movie_data_df = spark.createDataFrame(movie_data, columns)
movie_data_df.show()
```

```python
filter_movie_df = movie_data_df.filter("rating >= 8")
filter_movie_df.show()
```

```python
movie_df = filter_movie_df.select(
    "movie_name",
    "category"
)
movie_df.show()
```

```python
final_movie_df = filter_movie_df.withColumn(
    "collection_with_10%bonus",
    expr("collection_crore * 1.1")
)
final_movie_df.show()
```
