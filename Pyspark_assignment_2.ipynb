Question 6: RDD – Employee Salary Processing



```python
employee_data = [
    (101, "Arun", "IT", 45000),
    (102, "Divya", "HR", 38000),
    (103, "Kavin", "IT", 62000),
    (104, "Meena", "Finance", 55000),
    (105, "Ravi", "IT", 29000),
    (106, "Priya", "HR", 48000),
    (107, "Sanjay", "Finance", 72000),
    (108, "Nisha", "IT", 51000),
    (109, "Vijay", "HR", 32000),
    (110, "Anu", "Finance", 42000)
]

columns =["employee_id", "employee_name", "department", "salary"]

rdd_1 = spark.sparkContext.parallelize(employee_data)
rdd_1.collect()
```

```python
filter_rdd = rdd_1.filter(
    lambda x: x[2] == "IT" and x[3] > 40000
    )
filter_rdd.collect()
```

```python
filter_1_rdd = rdd_1.filter(
    lambda x: not (x[2] == "IT" and x[3] > 40000)
    )
filter_1_rdd.collect()
```

```python
import builtins
increment_rdd = filter_rdd.map(
    lambda x: (x[0], x[1], x[2],x[3],
               builtins.round(x[3] * 1.1, 3))
    )

increment_rdd.collect()
```

```python
final_rdd_emp = increment_rdd.map(
    lambda x: f"Employee: {x[1]} | Department: {x[2]} | Original_salary: {x[3]} | Revised_salary {x[4]}"
)
final_rdd_emp.collect()
```

Question 7: DataFrame – Student Result Processing

```python
student_data = [
    (1, "Arun", 85, 78, 92),
    (2, "Divya", 45, 55, 60),
    (3, "Kavin", 35, 40, 38),
    (4, "Meena", 90, 88, 95),
    (5, "Ravi", 65, 72, 58),
    (6, "Priya", 30, 85, 80),
    (7, "Sanjay", 55, 60, 65),
    (8, "Nisha", 95, 92, 89),
    (9, "Vijay", 40, 35, 70),
    (10, "Anu", 75, 80, 85)
]
columns = ["student_id", "name", "python", "sql", "pyspark"]

student_df = spark.createDataFrame(student_data,columns)
student_df.show()
```

```python
student_df_1 = student_df.withColumn("total_marks",
                                     expr("python +sql + pyspark")
                                     )
student_df_1.show()
```

```python
stud_df_2 = student_df_1.withColumn("average_marks",
                                     expr("round(total_marks/3, 3)")
                                     )
stud_df_2.show()
```

```python
stud_df_3 = stud_df_2.withColumn("result",
                                 when(col("python") >= 40, "Pass")
                                 .when(col("sql") >= 40, "Pass")
                                 .when(col("pyspark") >= 40, "Pass")
                                 .otherwise("Fail")
)
stud_df_3.show()
```

```python
stud_df_4 = stud_df_3.withColumn("grade",
                                 when(col("average_marks") >= 90, "A")
                                 .when(col("average_marks") >= 75, "B")
                                 .when(col("average_marks") >= 60, "C")
                                 .when(col("average_marks") <= 60, "D")
                                )
stud_df_4.show()
```

```python
stud_df_5 = stud_df_4.select(
    "name",
    "total_marks",
    "average_marks",
    "result",
    "grade"
)
stud_df_5.show()
```

```python
from pyspark.sql.functions import expr
final_result_df = stud_df_5.filter(
                                  expr("result == 'Pass' AND average_marks >= 75")
                                   )
final_result_df.show()
```
